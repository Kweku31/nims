```python
#importing the necessary libraries
import altair as alt
import pandas as pd
import seaborn as sns
```

```python
#load data from csv file
source = pd.read_csv('IHME-cardiovascular-deaths.csv')
```

```python
#read the file 
source.head(10)
```

```python
#Summary of data
print(source.info())

#Check for null values
print(F"\n{source.isnull().sum()}")
```

#Come up with your own analysis question to answer using the dataset.

What is the yearly percentage of deaths caused by Cardiovascular disaeases by location influenced by sex?

```python
#defining sex_age for ourc color coding
source['sex_age'] = source['sex'] + ' - ' + source['age']


# Generate shades automatically
unique_sex = source['sex'].unique()  # ['Male', 'Female']
unique_age = source['age'].unique()  # ['<5 years', '5-14 years']
n_age = len(unique_age)


# Create color palettes for Male (blues) and Female (oranges)
male_colors = sns.color_palette("Blues", n_colors=n_age).as_hex()  # Shades of blue
female_colors = sns.color_palette("Oranges", n_colors=n_age).as_hex()  # Shades of orange

# Map colors to sex_age combinations
color_domain = [f"{s} - {a}" for s in unique_sex for a in unique_age]
color_range = male_colors + female_colors  # Combine shades
```

```python

# Stacked bar chart
alt.Chart(source).mark_bar().encode(
    x='location:O',
    y=alt.Y('val:Q', scale=alt.Scale(zero=True), title='Percentage of Deaths'),
    color=alt.Color('sex_age:N', scale=alt.Scale(
        domain=color_domain,
        range=color_range
    ), title='Sex and Age'),
    column='year:N',
    tooltip=['sex', 'age', 'val', 'year', 'cause']
).properties(
    title='What is the yearly percentage of deaths caused by Cardiovascular disaeases by location?',
    width=100,
    height=700
).interactive()

```

The analysis question for the dataset was, What is the yearly percentage of deaths caused by Cardiovascular diseases by location? The rationale for this question was to determine how the percentage of deaths caused by Cardiovascular diseases varies by location each year. The answer to the question helps decide on the location that has seen an increase or decrease in its yearly percentage of deaths caused by Cardiovascular diseases.

The stacked bar chart was an ideal visualisation for the question since we can visualise for a location the aggregate or total percentage of deaths in a year caused by Cardiovascular Diseases, and can categorise the data by location. The visualisation can also help compare the location with the highest percentage of deaths and the location with the lowest percentage of deaths for a particular year.

The actual visualisation, I wanted to show how for each stacked bar for a  location in a year, we can identify the percentage of deaths for each sex by using a blue color scheme for the males and an orange color scheme for the females. After getting the stacked bar chart, I wanted to show that for each location, after knowing the percentage of death for each sex, I could make use of the ages given in the dataset to also describe the percentage of death for each sex categorized by age.

The data was read using pandas, and I preprocessed the data to check for null values since they can bring inconsistency to the visualisation, and checked for information about the dataset.
