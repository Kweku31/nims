```python
import pandas as pd
import altair as alt
from vega_datasets import data

# load datasets
counties = alt.topo_feature(data.us_10m.url, 'counties') # county-level US map data
states = alt.topo_feature(data.us_10m.url, 'states') # state-level US map data
waPopulation = pd.read_csv("WACountyPopulations.csv") # Washington state population counts

# process WA state population data
waPopulation[["Census2020", "Estimate2021", "Estimate2022"]] = waPopulation[
        ["Census2020", "Estimate2021", "Estimate2022"]
    ].apply(pd.to_numeric)
waPopulation = waPopulation.rename(columns={"GEOID_TIGER": "id"})

waPopulation
```
