```python
# Calculate total rows
total_rows = len(source)
```

```python
#plotting bar chart
barchart = alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender'] 
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender'] 
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage = 'datum.total/ datum.total * 100'
).mark_bar().encode(
    x=alt.X('Work_Arrangement:N', title='Work Arrangement'),
    y=alt.Y('percentage:Q', title='Percentage (%)'),  
    xOffset='Burnout_Level:N',  
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),  
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')

    ],
    column=alt.Column('Gender:N', title='Gender') 
).properties(
    title=alt.TitleParams('How do different work arrangements impact burnout for different genders?',
    fontWeight='bold',
    anchor='middle'),
    width=300,
    height=300
).interactive()



#plotting doughnut chart
doughnut = alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender']  
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender']
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage='datum.total/datum.total * 100'
).mark_arc(innerRadius=20).encode(  # innerRadius creates doughnut
    theta=alt.Theta('percentage:Q', title='Percentage (%)'),
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')
    ]
).properties(
    width=150,
    height=150
).facet(
    column=alt.Column('Work_Arrangement:N', title='Work Arrangement'),
    row=alt.Row('Gender:N', title='Gender'),
    spacing=10
).properties(
    title=alt.TitleParams(
        text='Burnout Distribution by Work Arrangement and Gender',
        fontWeight='bold',
        anchor='middle'
    )
).interactive()


alt.hconcat(barchart,doughnut)
```

## Analysis

The analysis question, which was "How do different work arrangements impact burnout for different genders?" was visualised using the bar chart with x-offset. The x-offset was the burnout level, the x-axis was the work arrangement, the y-axis was the percentage of the count of the burnout level, and the columns were gender. In the visualisation, the color code for the burnout level was inspired by the triage color code. Based on the visualization, we can say that females who work onsite have a higher burnout level compared to other genders who work onsite. And the male hybrid workers have a higher burnout level than the other gender who are hybrid workers.

From the second visualisation, based on the gender and the work arrangement, we can see the proportions of burnout levels among the workers. In the doughnut chart, the proportions are based on the work arrangements for each gender.

```python
#plotting doughnut chart
alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender']  
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender']
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage='datum.total/datum.total * 100'
).mark_arc(innerRadius=20).encode(  # innerRadius creates doughnut
    theta=alt.Theta('percentage:Q', title='Percentage (%)'),
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')
    ]
).properties(
    width=150,
    height=150
).facet(
    column=alt.Column('Work_Arrangement:N', title='Work Arrangement'),
    row=alt.Row('Gender:N', title='Gender'),
    spacing=10
).properties(
    title=alt.TitleParams(
        text='Burnout Distribution by Work Arrangement and Gender',
        fontWeight='bold',
        anchor='middle'
    )
).interactive()

```
