```python
All = ['India' ,'China' ,'France', 'Canada' ,'USA' ,'Argentina' ,'Australia', 'Nigeria','Russia', 'Brazil']
country = [All,'India' ,'China' ,'France', 'Canada' ,'USA' ,'Argentina' ,'Australia', 'Nigeria','Russia', 'Brazil']
# country = ['India' ,'China' ,'France', 'Canada' ,'USA' ,'Argentina' ,'Australia', 'Nigeria','Russia', 'Brazil']

base = alt.Chart(source).mark_line().encode(
    x='Year:N',
    y='average(Economic_Impact_Million_USD):Q',
    tooltip=[alt.Tooltip('Year:N', title='Year'), alt.Tooltip('average(Economic_Impact_Million_USD):Q', title='Average Economic Impact Million USD')] 

)

# A dropdown filter
country_dropdown = alt.binding_select(options=country, name="Country")
country_select = alt.selection_point(fields=['Country'], bind=country_dropdown)

filter_country = base.add_params(
    country_select
).transform_filter(
    country_select
).properties(title="Average Economic Impact(USD) by Country")


filter_country
```
