```python
import altair as alt
import pandas as pd
import seaborn as sns
import numpy as np

# Create DataFrame

df = pd.read_csv('movies_small.csv')
```

```python
df.head()
```

```python
alt.Chart(df).mark_bar().encode(
    x='Genre:O',
    y=alt.Y('Rotten_Tomatoes_Rating:Q', title='Rotten Tomatoes Rating'),
    color='Release_Year:N',
    # column='Release_Year:N'
).properties(
    title='How does movie genre affect Rotten Tomatoes Ratings?',
    width=100,
    height=300
).interactive()
```

```python
alt.Chart(df).mark_bar().encode(
    x='Title:N',
    y=alt.Y('Rotten_Tomatoes_Rating:Q', title='Rotten Tomatoes Rating'),
    color='Release_Year:N',
    # column='Release_Year:N'
).properties(
    title='Which movie has the highest rotten tomatoes rating?',
    width=800,
    height=300
).interactive()
```
