## Exploratory Data Analysis and Visualisation

1. **Bar Chart**: To visually compare the values of different categories or groups like the columns survey_date, gender, region, industry, job_role, work_arrangement, mental_health_status, and burnout_level, physical_health_issues.
1. **Histogram**: To visually indicate the frequency distribution or shape of numeric feature in the dataset like age, hours_per_week, Work_Life_Balance_Score, Social_Isolation_Score,and Salary_Range

```python
# Calculate total rows
total_rows = len(source)

alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender'] 
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender'] 
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage = 'datum.total/ datum.total * 100'
).mark_bar().encode(
    x=alt.X('Work_Arrangement:N', title='Work Arrangement'),
    y=alt.Y('percentage:Q', title='Percentage (%)'),  
    xOffset='Burnout_Level:N',  
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),  
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')

    ],
    column=alt.Column('Gender:N', title='Gender') 
).properties(
    title=alt.TitleParams('How do different work arrangements impact burnout for different genders?',
    fontWeight='bold',
    anchor='middle'),
    width=300,
    height=300
).interactive()
```

```python
# Calculate total rows
total_rows = len(source)

base = alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender']  
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender']
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage='datum.total/datum.total * 100'
).mark_arc(innerRadius=50).encode(  # innerRadius creates doughnut
    theta=alt.Theta('percentage:Q', title='Percentage (%)'),
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')
    ]
).properties(
    width=150,
    height=150
)

# Faceted chart
chart = base.facet(
    column=alt.Column('Work_Arrangement:N', title='Work Arrangement'),
    row=alt.Row('Gender:N', title='Gender'),
    spacing=10
).properties(
    title=alt.TitleParams(
        text='Burnout Distribution by Work Arrangement and Gender',
        fontWeight='bold',
        anchor='middle'
    )
).interactive()

# Display
chart
```

```python
base = alt.Chart(source).transform_aggregate(
    count='count()',
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender']
).transform_window(
    total='sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender']
).transform_calculate(
    percentage='datum.count / datum.total * 100',
    total_percentage='datum.total / datum.total * 100'
).encode(
    theta=alt.Theta('percentage:Q', stack=True),
    radius=alt.Radius('percentage:Q', scale=alt.Scale(type='linear', zero=True, rangeMin=20)),
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')
    ]
).properties(
    width=500,
    height=300
)

# Radial bars and text labels
c1 = base.mark_arc(innerRadius=20, stroke='#fff')
c2 = base.mark_text(radiusOffset=10).encode(text=alt.Text('percentage:Q', format='.2f'))

# Faceted chart
chart = (c1 + c2).facet(
    column=alt.Column('Work_Arrangement:N', title='Work Arrangement'),
    row=alt.Row('Gender:N', title='Gender'),
    columns=3
).properties(
    title=alt.TitleParams(
        text='Burnout Distribution by Work Arrangement and Gender',
        fontWeight='bold',
        anchor='middle'
    )
).interactive()

# Display
chart
```

```python
mental_filtered = source[source['Mental_Health_Status'].isin(['Depression', 'ADHD'])]

# Extract the upper bound (or lower bound if "$XK+")
source['sort_key'] = (
    source['Salary_Range']
    .str.extract(r'\$(\d+)K')  # Extract first number after "$" and before "K"
    .astype(float) * 1000      # Convert to numeric (e.g., 40K → 40000)
)

# Sort the DataFrame from highest to lowest
salary_sorted = source.sort_values('sort_key', ascending=False)

# Define salary ranges in descending order
salary_order = ['$120K+', '$100K-120K', '$80K-100K', '$60K-80K', '$40K-60K']
# Define colors
salary_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#9467bd', '#d62728']


chart = alt.Chart(mental_filtered).mark_bar().encode(
    x=alt.X('Salary_Range:N', title='Salary Range', sort=alt.EncodingSortField(field='sort_key', order='ascending')),
    y=alt.Y('count():Q', title='Number of People'),
    color=alt.Color('Salary_Range:N', scale=alt.Scale(domain=salary_order, range=salary_colors)),
    column=alt.Column('Industry:N', title='Industry'),
    row=alt.Row('Mental_Health_Status:N', title='Mental Health Status'),
    tooltip=[
        alt.Tooltip('Industry:N'),
        alt.Tooltip('Mental_Health_Status:N'),
        alt.Tooltip('Salary_Range:N'),
        alt.Tooltip('count():Q', title='Count')
    ]
).properties(
    title='Distribution of Salary by Industry and Mental Health Status',
    width=70,  
    height=200
).configure_axisX(labelAngle=-45)

chart
```
