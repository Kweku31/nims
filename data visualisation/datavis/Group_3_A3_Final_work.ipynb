### Preparing the variables to be used 

```python
# Disable row limit
alt.data_transformers.disable_max_rows()

# Dropdown options
years = sorted(source['Year'].unique())
crops = sorted(source['Crop_Type'].unique())
countries = sorted(source['Country'].unique())

# Params for dropdown filters

#dropdown for years
year_dropdown = alt.binding_select(options=years, name='Year of the mean crop yield: ')
year_param = alt.param(name='year_param', bind=year_dropdown, value=years[0])  

#dropdown for crops
crop_dropdown = alt.binding_select(options=crops, name='Crop Type: ')
crop_param = alt.param(name='crop_param', bind=crop_dropdown, value=crops[0])

#dropdown for country
country_dropdown = alt.binding_select(options=countries, name='Country: ')
country_param = alt.param(name='country_param', bind=country_dropdown, value=countries[0])

# Determine greeting based on current time
current_hour = datetime.now().hour
if current_hour < 12:
    greeting = "Good Morning"
elif current_hour < 18:
    greeting = "Good Afternoon"
else:
    greeting = "Good Evening"


# Custom color scale for Crop_Type
crop_colors = alt.Scale(
    domain=["Corn", "Wheat", "Coffee", "Sugarcane", "Fruits", "Rice", "Barley", "Vegetables", "Soybeans", "Cotton"],
    range=[
        "#1f77b4",  # Corn: Blue
        "#ff7f0e",  # Wheat: Orange
        "#8c564b",  # Coffee: Brown
        "#d62728",  # Sugarcane: Red
        "#9467bd",  # Fruits: Purple
        "#2ca02c",  # Rice: Green
        "#e377c2",  # Barley: Pink
        "#7f7f7f",  # Vegetables: Gray
        "#bcbd22",  # Soybeans: Olive
        "#17becf"   # Cotton: Cyan
    ]
)
```

### The charts for the various components to be used on the dashboard

```python
# -------------------
# Chart 1: Area chart of the mean pesticide used KG per HA across the years filtered by the country and crop type
# -------------------
chart1 = (
    alt.Chart(source)
    .mark_area(
        line={'color':'darkgreen'},
    color=alt.Gradient(
        gradient='linear',
        stops=[alt.GradientStop(color='white', offset=0),
               alt.GradientStop(color='darkgreen', offset=1)],
        x1=1,
        x2=1,
        y1=1,
        y2=0
    )
    )
    .encode(
        x=alt.X('Year:O', title='Year'),
        y=alt.Y('mean(Pesticide_Use_KG_per_HA):Q', title='Mean Pesticide_Use_KG_per_HA'),
        color=alt.Color('Crop_Type:N', scale=crop_colors, title='Crop Type'),
        tooltip=['Crop_Type', 'Country', 'Year', alt.Tooltip('mean(Pesticide_Use_KG_per_HA):Q', title='Average Pesticide_Use_KG_per_HA')]
    ).transform_filter(
        (alt.datum.Crop_Type == crop_param) 
    ).transform_filter(
        (alt.datum.Country == country_param)
    ).add_params(
        crop_param, country_param
    ).properties(
        title='Distribution of the mean Pesticide_Use_KG_per_HA'
    )
)

# -------------------
# Chart 2: Line chart for mean Economic Impact filtered by the country and crop type
# -------------------
chart2 = (
    alt.Chart(source)
    .mark_line(point=True)
    .encode(
        x=alt.X('Year:O', title='Year'),
        y=alt.Y('mean(Economic_Impact_Million_USD):Q', title='Mean Economic Impact (Million USD)'),
        color=alt.Color('Crop_Type:N'),
        # fill=alt.Fill('Crop_Type:N'),
        tooltip=['Year', 'Country', 'Crop_Type', alt.Tooltip('mean(Economic_Impact_Million_USD):Q', title='Mean Economic Impact')]
    ).transform_filter(
        (alt.datum.Crop_Type == crop_param) 
    ).transform_filter(
        (alt.datum.Country == country_param) 
    ).add_params(
         crop_param, country_param
    ).properties(title='Mean Economic Impact Over Time')
)


# -------------------
# Chart 3: Donut chart for mean crop yield MT per HA filtered by the country and crop type 
# and the years and the proportion representing the regions for the country
# -------------------

#selection parameter to enable selecting the region of the donut chart
region_selector = alt.selection_point(fields=["Region"], on="click", encodings=['fill'])

#donut chart body
chart3 = alt.Chart(source).mark_arc(innerRadius=50).encode(
    theta=alt.Theta(field="Crop_Yield_MT_per_HA", aggregate="mean", type="quantitative"),
    # color=alt.when(region_selector).then("Region:N").otherwise(alt.value("lightgray")),
    fill=alt.Fill("Region:N", title="Region"),
    tooltip=[
        alt.Tooltip('Year:N', title='Year'),
        alt.Tooltip('Region:N', title='Region'),
        alt.Tooltip('Crop_Yield_MT_per_HA:Q', aggregate='mean', title='Mean Crop Yield (MT/HA)'),
        alt.Tooltip('Crop_Type:N', title='Crop Type'),
        alt.Tooltip('Country:N', title='Country')
    ]
).transform_filter(
    (alt.datum.Crop_Type == crop_param) 
).transform_filter(
    (alt.datum.Country == country_param)
).add_params(
    crop_param, country_param, year_param, region_selector
).properties(
    width=300,
    height=300,
    title=alt.TitleParams(
        text="Mean Crop Yield",
        fontSize=14
    )
).transform_filter(
    (alt.datum.Year == year_param) 
)

# -------------------
# Chart 4: Bar chart for adaptation strategies filtered by the regions for the country when clicked by the 
# region on the donut chart and the year
# -------------------
chart4 = alt.Chart(source).mark_bar().encode(
    x=alt.X("count()",title='Count of Adaptation Strategies'),
    y=alt.Y("Adaptation_Strategies:N").sort("-x"),      
    tooltip=["count()"],
).transform_filter(
    (alt.datum.Crop_Type == crop_param) 
).transform_filter(
    (alt.datum.Country == country_param) 
).transform_filter(
    (alt.datum.Year == year_param) 
).transform_filter(
    region_selector 
).add_params(
    crop_param, country_param, year_param, region_selector
).properties(
    height=200,
)

# -------------------
# Chart 5: Bar chart for count of records filtered by the country and crop type for the years 
# -------------------
chart5 = (
    alt.Chart(source)
    .mark_bar()
    .encode(
        x=alt.X('Year:O', title='Year'),
        y=alt.Y('count():Q', title='Count of Crop'),
        color=alt.Color('Crop_Type:N'),
        tooltip=['Crop_Type', 'Country', 'Year', alt.Tooltip('count():Q', title='Count')]
    ).transform_filter(
        (alt.datum.Crop_Type == crop_param) 
    ).transform_filter(
        (alt.datum.Country == country_param) 
    ).add_params(
        crop_param, country_param
    ).properties(
        title='Distribution of Crop Types in Different Countries by Year'
    )
)


# Combine horizontally and vertically and apply axis config
final_chart = (
    alt.hconcat(alt.vconcat(chart1,alt.hconcat(chart3,chart4)),alt.vconcat(chart5,chart2))
    .resolve_scale(color='shared')
    .configure_axisX(labelAngle=-45)
    .properties(
        title=alt.TitleParams(
        text=f"{greeting}, Welcome To Your Agriculture Dashboard",
        fontWeight='bolder',
        fontSize = 20,
        anchor='middle'
    )
    )
)

#uncomment the line below to see the dashboard
# final_chart
```

## Design Rationale

Our Interactive Dashboard was made to analyze the impact of the various factors in Agriculture from the years 1990 to 2024. The dataset consisted of Year of the recorded data, the Country from which the data was recorded as well as the specific region in the Country, the Average Temperature in Celcius, Total Precipitation in milligrams,Pesticide use in Kilograms per hector amongst others.
Our Target Audience involves a wide range of individuals as well as organizations.They include Agricultural policymakers and government agencies,Climate and environmental researchers,Agribusiness stakeholders and farmers,NGOs implementing climate adaptation programs as well as Individauls interested in the welfare of Economic Development in Agriculture.

We are however motivated by the large range of our target audience to research on  Crop yields, Climate and Economic impact relationships, enable filtering by Crop type, Country and Year for targeted insights, and support Data driven agricultural Decision making.



Our interactive dashboard features four visualizations, each filterable by Crop Type, Country, and Year of Mean Crop Yield. The visualizations are as follows:


1. Area Graph(Visualization 1): This visualization illustrates the mean pesticide usage (in kilograms per hectare) across the years 1990–2024, highlighting temporal trends.


2. Bar Graph(Visualization 2): This chart plots the count of various crops (e.g., Corn, Barley, Coffee, Cotton) against the years 1990–2024, showing crop distribution over time.


3. Donut Chart(Visualization 3): This visualization displays the mean crop yield across different regions in various countries, providing a clear regional comparison.


4. Horizontal Bar Chart(Visualization 4): This chart showcases the count of adaptation strategies, emphasizing their prevalence.


5. Line Chart(Visualization 5): This chart plots the Mean Economic Impact (in million dollars) over the years 1990–2024, illustrating economic trends over time.



The visual encodings that were used to ensure Effectiveness in our visualisations were Position, Length, Color Hue and Area from the Cleveland McGill's principles of Graphical perceptions and the Relative Magnitude Comparison.

Cleveland and McGill's principles helps provide a framework for designing effective visualizations by ranking visual encodings according to how accurately humans perceive and compare data and The relative magnitude comparison in perception helps viewers interpret the size, intensity, or value of visual elements (e.g., bars, lines, or colors) by comparing them to other elements within the visualization, rather than assessing them in absolute terms. This perceptual process significantly influences how data is understood.

Visualisations 1,2and 5 are being plotted with  quantitative variables in both axes. The visual encodings used in our visualisations were Position, Length and Area. Visualisation 3 and 4 however, was plotted with a nominal and  a quantitative variable and  Angle as well as Colour Hue .Included also, were tooltips which improves usability and user experience by clarifying functionality.


Our Chosen Interactions enable our Target Audience;
1. Provide insights on Crop yield
2. Provides insights for Research manangement and Market Trends.
3. Increase Productivity and Reduce costs.
4. Make better-informed decisions 

All of this ultimately improving livelihood and food security.
