## **Question 1** :  How does the salary of workers  vary across different Industries and mental health statuses?

```python
#Bar chart for Question 1
# Extract the upper bound (or lower bound if "$XK+")
source['sort_key'] = (
    source['Salary_Range']
    .str.extract(r'\$(\d+)K')  # Extract first number after "$" and before "K"
    .astype(float) * 1000      # Convert to numeric (e.g., 40K → 40000)
)

# Sort the DataFrame from highest to lowest
salary_sorted = source.sort_values('sort_key', ascending=False)

# Define salary ranges in descending order
salary_order = ['$120K+', '$100K-120K', '$80K-100K', '$60K-80K', '$40K-60K']
# Define colors
salary_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#9467bd', '#d62728']


chart_1 = alt.Chart(source).mark_bar().encode(
    x=alt.X('Salary_Range:N', title='Salary Range', sort=alt.EncodingSortField(field='sort_key', order='ascending')),
    y=alt.Y('count():Q', title='Number of People'),
    color=alt.Color('Salary_Range:N', scale=alt.Scale(domain=salary_order, range=salary_colors)),
    column=alt.Column('Industry:N', title='Industry'),
    row=alt.Row('Mental_Health_Status:N', title='Mental Health Status'),
    tooltip=[
        alt.Tooltip('Industry:N'),
        alt.Tooltip('Mental_Health_Status:N'),
        alt.Tooltip('Salary_Range:N'),
        alt.Tooltip('count():Q', title='Count')
    ]
).properties(
    title='Distribution of Salary by Industry and Mental Health Status',
    width=70,  
    height=200
).configure_axisX(labelAngle=-45)

chart_1
```

```python
#Heat map for Question 1
chart = alt.Chart(source).mark_rect().encode(
    x=alt.X('Salary_Range:N', title='Salary Range'),
    y=alt.Y('Industry:N', title='Industry'),
    color=alt.Color('count():Q', title='Number of People'),
    row=alt.Row('Mental_Health_Status:N', title='Mental Health Status'),
    tooltip=[
        alt.Tooltip('Industry:N'),
        alt.Tooltip('Mental_Health_Status:N'),
        alt.Tooltip('Salary_Range:N'),
        alt.Tooltip('count():Q', title='Count')
    ]
).properties(
    title='Heatmap of Salary Distribution by Industry and Mental Health Status',
    width=150,
    height=150
).configure_axisX(labelAngle=-45)

chart
```

### Analysis

To address our first sub-question, we created a bar chart and a heatmap, where each color represented a different salary range. The visualization mapped various mental health statuses (e.g., ADHD, anxiety, burnout, or none) across different industries (e.g., Customer Service, Education, etc.), with salary ranges such as $60K–$80K, $80K–$100K, $100K–$120K, and $120K+ indicated by color.

Our analysis revealed that most employees across industries earned salaries within the $60K–$80K and $80K–$100K ranges. Notably, individuals earning between $60K–$80K had the highest prevalence of reported mental health issues, including anxiety and burnout. From this, we concluded that lower salary ranges were generally associated with a higher incidence of mental health challenges among employees, and conversely, higher salary ranges correlated with fewer reported mental health issues.

## **Question 2**: How do different work arrangements impact burnout for individuals of different genders? 


```python
#plotting bar chart
barchart = alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender'] 
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender'] 
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage = 'datum.total/ datum.total * 100'
).mark_bar().encode(
    x=alt.X('Work_Arrangement:N', title='Work Arrangement'),
    y=alt.Y('percentage:Q', title='Percentage (%)'),  
    xOffset='Burnout_Level:N',  
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),  
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')

    ],
    column=alt.Column('Gender:N', title='Gender') 
).properties(
    title=alt.TitleParams('How do different work arrangements impact burnout for different genders?',
    fontWeight='bold',
    anchor='middle'),
    width=300,
    height=300
).interactive()



#plotting doughnut chart
doughnut = alt.Chart(source).transform_aggregate(
    count='count()',  
    groupby=['Work_Arrangement', 'Burnout_Level', 'Gender']  
).transform_window(
    total = 'sum(count)',
    frame=[None, None],
    groupby=['Work_Arrangement', 'Gender']
).transform_calculate(
    percentage=f'datum.count / {total_rows} * 100',
    total_percentage='datum.total/datum.total * 100'
).mark_arc(innerRadius=20).encode(  # innerRadius creates doughnut
    theta=alt.Theta('percentage:Q', title='Percentage (%)'),
    color=alt.Color('Burnout_Level:N', scale=alt.Scale(domain=['Low', 'Medium', 'High'], range=['#2ca02c', '#e8e85d', '#b21212']), title='Burnout Level'),
    tooltip=[
        alt.Tooltip('Work_Arrangement:N', title='Work Arrangement'),
        alt.Tooltip('Burnout_Level:N', title='Burnout Level'),
        alt.Tooltip('count:Q', title='Count'),
        alt.Tooltip('percentage:Q', title='Percentage', format='.2f'),
        alt.Tooltip('total_percentage:Q', title='Total Percentage', format='.2f')
    ]
).properties(
    width=150,
    height=150
).facet(
    column=alt.Column('Work_Arrangement:N', title='Work Arrangement'),
    row=alt.Row('Gender:N', title='Gender'),
    spacing=10
).properties(
    title=alt.TitleParams(
        text='Burnout Distribution by Work Arrangement and Gender',
        fontWeight='bold',
        anchor='middle'
    )
).interactive()


alt.hconcat(barchart,doughnut)
```

## Analysis

The analysis question, which was "How do different work arrangements impact burnout for different genders?" was visualised using the bar chart with x-offset. The x-offset was the burnout level, the x-axis was the work arrangement, the y-axis was the percentage of the count of the burnout level, and the columns were gender. In the visualisation, the color code for the burnout level was inspired by the triage color code. Based on the visualization, we can say that females who work onsite have a higher burnout level compared to other genders who work onsite. And the male hybrid workers have a higher burnout level than the other gender who are hybrid workers.

From the second visualisation, based on the gender and the work arrangement, we can see the proportions of burnout levels among the workers. In the doughnut chart, the proportions are based on the work arrangements for each gender.

## **Question 3**: Does the region of the work arrangement affect the mental health status of the various industries? 	

```python
# Stacked bar chart
alt.Chart(source).mark_bar().encode(
    x=alt.X('Industry:N', title='Industry'),  
    y=alt.Y('count():Q', title='Count of mental_health_status'),  
    xOffset='Mental_Health_Status:N',  
    color=alt.Color('Mental_Health_Status:N', scale=alt.Scale(domain=['Stress Disorder', 'ADHD', 'None', 'Burnout', 'Anxiety', 'PTSD', 'Depression'], range=["#ffd700","#ffb14e","#fa8d15","#ed2571","#1de8f3","#9d02d7","#0000ff"] ), title='Mental_Health_Status'),  
    tooltip=[
        alt.Tooltip('Region:N', title='Region'),
        alt.Tooltip('Mental_Health_Status:N', title='Mental Health Status'),
        alt.Tooltip('count():Q', title='Count')
    ]).properties(
     width=600,
    height=300
).facet(
    facet='Region:N',  # Use facet instead of column for flexibility
    columns=2,  # Set 2 regions per row
    title='Region',
    spacing=10 # Adjust spacing between facets for alignment
).properties(
    title='Does the Region of the Work Arrangement affect the Mental Health Status for various Industries?'
).interactive()
```
