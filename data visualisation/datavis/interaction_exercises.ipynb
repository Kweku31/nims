### Visualization 1

```python
alt.Chart(movies).mark_point().encode(
    x=alt.X("US_Gross:Q").scale(type="log"),
    y=alt.Y("Worldwide_Gross:Q").scale(type="log")
).transform_calculate(
    WorldMinusUSGross=alt.datum.Worldwide_Gross-alt.datum.US_Gross
).transform_filter(
    "datum.US_Gross > 0 && datum.WorldMinusUSGross > 0"
)
```

### Visualization 2

```python
alt.Chart(movies).mark_bar().encode(
    x=alt.X("Major_Genre:N").sort("-y"),
    y=alt.Y("average(US_Gross):Q")
)
```

### Visualization 3

```python
alt.Chart(movies).mark_area().encode(
    x=alt.X("year(Release_Date):T"),
    y=alt.Y("count():Q")
)
```

```python
# date_range = (dt.date(2007, 6, 30), dt.date(2009, 6, 30))


#Interaction visualisation
brush = alt.selection_interval(encodings=['x']
                            #    ,value={'x': Release_Date}
                               )

base = alt.Chart(movies).mark_area().encode(
    x=alt.X("year(Release_Date):T"),
    y=alt.Y("count():Q"),
    tooltip=[
        alt.Tooltip('year(Release_Date):T', title='year(Release_Date)'),
        alt.Tooltip('count():Q', title='Count')
    ]
).properties(
    width=600,
    height=200
)

upper = base.encode(
    alt.X('year(Release_Date):T', scale=alt.Scale(domain=brush))
)

lower = base.properties(
    height=200
).add_params(brush)

upper & lower
```
